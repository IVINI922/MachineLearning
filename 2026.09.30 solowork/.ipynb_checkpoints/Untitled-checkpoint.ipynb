# Описание модели машинного обучения

## Алгоритм: TransformedTargetRegressor поверх HistGradientBoostingRegressor (scikit-learn) — гистограммный градиентный бустинг над решающими деревьями с автоматическим логарифмированием целевой переменной.

#### Как работает HistGradientBoosting:

Старт. Модель начинается со среднего log(price).

Итерации. На каждом шаге:

считаются остатки — разница между реальной и текущей ценой;

обучается новое дерево на остатках;

дерево добавляется в ансамбль с весом learning_rate.

Особенность. Признаки бинаризуются в гистограммы (до 255 бинов) — это сильно ускоряет обучение и уменьшает память по сравнению с обычным Gradient Boosting.

Выход. TransformedTargetRegressor делает log1p/expm1 автоматически → цена в TRY.

```python
import json
meta = json.load(open("artifacts/model_metadata.json", encoding="utf-8"))
for k, v in meta["metrics"].items():
    print(f"{k:<12} = {v:.4f}")
```

R² (коэффициент детерминации) — доля объяснённой дисперсии. От 0 до 1, чем ближе к 1 — тем лучше.

R²_log = 0.910 — модель объясняет 91% разброса логарифма цены.

R²_try = 0.929 — в исходной шкале (в TRY) объясняет 93%.

MAE (Mean Absolute Error) — средняя абсолютная ошибка.

MAE_log = 0.078 — в среднем модель ошибается на 0.078 в лог-шкале.

RMSE (Root Mean Squared Error) — корень из средней квадратичной ошибки. Сильнее штрафует большие ошибки.

RMSE_log = 0.107

MAPE (Mean Absolute Percentage Error) — средняя ошибка в процентах. Самая понятная метрика.

MAPE = 7.92% — модель ошибается в среднем на 8% от стоимости объекта.

Для квартиры за 5 млн TRY → ошибка ≈ ±400 000 TRY.

# Проверка модели на 20 тестовых объектах

```python
import json
import pickle

import joblib
import numpy as np
import pandas as pd

# --- настройка вывода: пробелы вместо экспоненты ---
pd.set_option("display.float_format", lambda x: f"{x:,.0f}")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", 40)

# ---------------- Пути ----------------
ART = "artifacts"
TEST_PATH = "test_real_estate_samples.csv"
RESULT_PATH = "result_real_estate_samples.csv"

# ---------------- Загрузка артефактов ----------------
model  = joblib.load(f"{ART}/final_model_gb.joblib")
scaler = joblib.load(f"{ART}/scaler.joblib")
te     = pickle.load(open(f"{ART}/target_encoders.pkl", "rb"))
feats  = json.load(open(f"{ART}/feature_names.json", encoding="utf-8"))

FEATURES = feats["final_features"]
SCALE_COLS = list(scaler.feature_names_in_)

# нормализация ключей target-энкодеров
for k, enc in te.items():
    if isinstance(enc, dict):
        te[k] = {str(kk).strip(): float(vv) for kk, vv in enc.items()}


# ---------------- Утилиты ----------------
def te_lookup(encoder, value):
    if not isinstance(encoder, dict):
        return 0.0
    v = str(value).strip() if value is not None else ""
    if v in encoder:
        return float(encoder[v])
    for k, val in encoder.items():
        if str(k).strip() == v:
            return float(val)
    vals = [x for x in encoder.values() if isinstance(x, (int, float))]
    return float(np.mean(vals)) if vals else 0.0


def parse_size(value):
    try:
        v = float(value)
    except (TypeError, ValueError):
        return 100.0
    if np.isnan(v) or v <= 0:
        return 100.0
    return v


def parse_age(value):
    if value is None or (isinstance(value, float) and pd.isna(value)):
        return 0.0
    s = str(value).strip()
    if s.isdigit():
        return float(s)
    return 0.0


def clamp_tom(value, tmax=23):
    try:
        return float(min(max(float(value), 0.0), tmax))
    except (TypeError, ValueError):
        return 0.0


def build_row(raw: dict) -> pd.DataFrame:
    row = {}
    row["size"] = parse_size(raw.get("size"))
    row["building_age"] = parse_age(raw.get("building_age"))
    row["tom"] = clamp_tom(raw.get("tom"))

    row["city_te"]             = te_lookup(te["city_target_encoder"],             raw.get("city"))
    row["room_count_te"]       = te_lookup(te["room_target_encoder"],             raw.get("room_count"))
    row["district_grouped_te"] = te_lookup(te["district_target_encoder"],         raw.get("district_grouped"))

    row["sub_type_New"]       = int(raw.get("sub_type") == "New")
    row["sub_type_Secondary"] = int(raw.get("sub_type") == "Secondary")
    row["heating_type_Central"]    = int(raw.get("heating_type") == "Central")
    row["heating_type_Individual"] = int(raw.get("heating_type") == "Individual")
    row["heating_type_None"]       = int(raw.get("heating_type") == "None")

    row["price_currency_RUB"] = 1
    row["price_currency_USD"] = 0

    X = pd.DataFrame([row]).reindex(columns=FEATURES, fill_value=0).fillna(0.0)

    X_scaled = X.copy()
    X_scaled[SCALE_COLS] = scaler.transform(X[SCALE_COLS].to_numpy())
    X_scaled = X_scaled.fillna(0.0)
    return X_scaled


# ---------------- Загрузка теста ----------------
df = pd.read_csv(TEST_PATH, low_memory=False)
print(f"Загружено объектов: {len(df)}")

parts = df["address"].astype("string").str.split("/", n=2, expand=True)
df["city"] = parts[0].str.strip()
df["district_grouped"] = parts[1].str.strip() if parts.shape[1] > 1 else None

# ---------------- Прогон модели ----------------
preds_log = []
for _, r in df.iterrows():
    X = build_row(r.to_dict())
    preds_log.append(float(model.predict(X)[0]))

df["predicted_price"] = np.expm1(preds_log).round().astype(int)


# ============================================================
# ТАБЛИЦА ДЛЯ ОТЧЁТА — часть 1: основные поля + прогноз
# ============================================================
print("\n" + "=" * 100)
print("ТАБЛИЦА 1: ПРОГНОЗ МОДЕЛИ")
print("=" * 100)

table1 = df[["id", "type", "sub_type", "listing_type",
             "room_count", "size", "address",
             "price", "predicted_price"]].copy()
table1["price"] = table1["price"].round().astype(int)

print(table1.to_string(index=False))


# ============================================================
# ТАБЛИЦА ДЛЯ ОТЧЁТА — часть 2: подробности + прогноз
# ============================================================
print("\n" + "=" * 100)
print("ТАБЛИЦА 2: ПОЛНАЯ ТАБЛИЦА С ПРОГНОЗОМ (последний столбец)")
print("=" * 100)

table2 = df.copy()
table2["price"] = table2["price"].round().astype(int)
# оставляем все исходные колонки + predicted_price в конце
cols_order = [c for c in df.columns if c not in ["city", "district_grouped", "predicted_price"]]
cols_order += ["predicted_price"]
table2 = table2[cols_order]

print(table2.to_string(index=False))


# ============================================================
# Сохранение
# ============================================================
df.to_csv(RESULT_PATH, index=False, encoding="utf-8-sig")
print(f"\nСохранено: {RESULT_PATH}")
```

# Метрики MAE, RMSE, R² — вручную и через sklearn

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

pd.set_option("display.float_format", lambda x: f"{x:.2f}")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

# ============================================================
# Ошибки по каждому объекту
# ============================================================
df["error"]         = df["predicted_price"] - df["price"]
df["abs_error"]     = df["error"].abs()
df["squared_error"] = df["error"] ** 2
df["pct_error"]     = df["abs_error"] / df["price"] * 100

# ============================================================
# Метрики — вручную
# ============================================================
y_true = df["price"].values
y_pred = df["predicted_price"].values

mae_manual  = np.mean(np.abs(y_true - y_pred))
rmse_manual = np.sqrt(np.mean((y_true - y_pred) ** 2))
r2_manual   = 1 - np.sum((y_true - y_pred) ** 2) / np.sum((y_true - y_true.mean()) ** 2)
mape_manual = np.mean(np.abs((y_true - y_pred) / y_true)) * 100

# ============================================================
# Метрики — через sklearn
# ============================================================
mae_sk  = mean_absolute_error(y_true, y_pred)
rmse_sk = np.sqrt(mean_squared_error(y_true, y_pred))
r2_sk   = r2_score(y_true, y_pred)

# ============================================================
# Таблица метрик (для вывода и сохранения)
# ============================================================
metrics_df = pd.DataFrame({
    "Метрика": ["MAE", "RMSE", "R2", "MAPE_pct"],
    "Вручную": [mae_manual, rmse_manual, r2_manual, mape_manual],
    "Sklearn": [mae_sk, rmse_sk, r2_sk, np.nan],
})

# ============================================================
# Сохранение: объекты + метрики в один CSV (без пустых строк)
# ============================================================
RESULT_PATH = "result_real_estate_samples.csv"

with open(RESULT_PATH, "w", encoding="utf-8-sig", newline="") as f:
    df.to_csv(f, index=False, float_format="%.2f")
    f.write("\n")
    metrics_df.to_csv(f, index=False, float_format="%.2f")

print(f"Сохранено: {RESULT_PATH}")

# ============================================================
# Интерактивный вывод
# ============================================================
print("\n=== ОБЪЕКТЫ ===")
display(df)

print("\n=== МЕТРИКИ ===")
display(metrics_df)
```

id — уникальный идентификатор объекта из тестового файла

listing_type — тип сделки: 1 — продажа, 2 — аренда

size — площадь объекта, м²

room_count — количество комнат в формате n+m (комнаты + санузлы)

price_real — реальная цена объекта из тестового файла, TRY

predicted_price — цена, спрогнозированная моделью, TRY

error — ошибка прогноза: predicted_price − price_real (со знаком)

abs_error — абсолютная ошибка: |error| (без знака)

squared_error — квадрат ошибки: error² (для расчёта RMSE и R²)

pct_error_pct — относительная ошибка в процентах: abs_error / price_real × 100

# Анализ результатов

Характеристика выборки. 20 объектов: 10 — продажа (listing_type=1), 10 — аренда (listing_type=2). Цены продажи: 103 600–2 450 000 TRY. Цены аренды: 750–19 000 TRY.

Результат. Модель показала низкое качество:

Метрика	Значение

MAE	≈ 15,5 млн TRY

RMSE	≈ 17 млн TRY

R²	отрицательный

MAPE	> 100 000 %

Все прогнозы лежат в диапазоне 13–19 млн TRY, реальные цены — в 10–5000 раз меньше. Для арендного объекта с ценой 750 TRY модель дала ≈ 15,6 млн TRY.

### Причины:

#### Рассинхрон артефактов. 
Модель обучена на синтетических данных (Moscow, size ~ N(60, 20), цены 10–20 млн), а target_encoders.pkl собран на реальном турецком датасете.

#### Несовпадение категорий. 
sub_type и heating_type в тесте — реальные (Daire, Fancoil, Yok), а модель знает только New/Secondary и Central/Individual/None. Все One-Hot-колонки = 0.

#### Смешение продажи и аренды. 
10 из 20 объектов — аренда. Модель обучена только на продажах, поэтому для аренды прогноз завышен на 3–4 порядка.

#### Пропуски. 
size = NaN (3 объекта), size = 1.0 (2 объекта, вероятно опечатки), room_count = "+".

#### Совпадение расчётов. 
Метрики вручную и через sklearn совпадают — расчёт корректен.

#### Вывод. 
Модель неприменима для реального прогноза. Требуется переобучение на real_estate_data.csv с сохранением всех артефактов из одного запуска. Отрицательные метрики — не ошибка расчёта, а объективная оценка: модель хуже константного прогноза (среднего).
